# High-Resolution Mermaid Diagrams for Team Project

This notebook takes text instructions for Mermaid diagrams and renders them as high-resolution images within the output cells. It uses an external rendering service (`mermaid.ink`) to generate the images.

## Setup

Run the following cell to define the rendering function.

```python
import json
import base64
from IPython.display import Image, display

def render_mermaid(mermaid_code, theme='default', bg_color='white'):
    """
    Renders a Mermaid diagram from text instructions using mermaid.ink.

    Args:
        mermaid_code (str): The Mermaid diagram syntax.
        theme (str): The theme for the diagram (e.g., 'default', 'forest', 'dark', 'neutral').
        bg_color (str): The background color of the image (e.g., 'white', 'transparent').
    """
    try:
        # Use json.dumps to properly construct and escape the JSON payload
        payload_dict = {
            "code": mermaid_code,
            "mermaid": {
                "theme": theme,
                "themeVariables": {
                    "background": bg_color
                }
            }
        }
        payload_json = json.dumps(payload_dict)

        # Encode the payload to base64 safely
        base64_bytes = base64.b64encode(payload_json.encode('utf-8'))
        base64_string = base64_bytes.decode('utf-8')

        # Construct the URL for the image request
        # Using the /img endpoint provides a static image
        url = f"https://mermaid.ink/img/{base64_string}"

        # Display the image in the notebook cell
        display(Image(url=url))

    except Exception as e:
        print(f"An error occurred while rendering the diagram: {e}")
```
